```python
!pip install --upgrade -q google-generativeai ipywidgets

import time
import google.generativeai as genai
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
def get_ai_response(prompt, api_key):
    if not api_key.strip():
        return "❌ Error: Please enter a valid Gemini API Key first."
    try:
        # Configure the library with your key
        genai.configure(api_key=api_key.strip())

        # Using 'gemini-1.5-flash' explicitly via the correct v1 model path
        # to guarantee compatibility across all environment versions.
        model = genai.GenerativeModel('gemini-2.5-flash')

        response = model.generate_content(prompt)
        return response.text
    except Exception as e:
        return f"❌ An error occurred: {str(e)}\n\nTip: Double check that your API key is correct and active."
```

```python
# Main Header Widget
title_html = widgets.HTML("<h1 style='color: #FF8F00; font-family: Arial;'>📚 Study-Buddy(make study easier..)</h1><p style='color: #555;'><i>Your direct Colab companion for explaining concepts, summarizing notes, and generating quizzes.</i></p><hr>")

# Config Elements
api_key_input = widgets.Password(description="API Key:", placeholder="Paste Gemini API Key here")
help_text = widgets.HTML("<small><a href='https://aistudio.google.com/' target='_blank'>Get a free Gemini API key here</a></small><br><br>")

# Feature 1: Explain Concept
concept_text = widgets.Text(description="Topic:", placeholder="e.g., Quantum Computing")
style_dropdown = widgets.Dropdown(options=["Explain like I'm 5", "Analogy-based", "Detailed Breakdown"], description="Style:")
btn_explain = widgets.Button(description="Explain Concept", button_style="primary")

# Feature 2: Summarize Notes
notes_text = widgets.Textarea(placeholder="Paste your textbook chapters or rough study notes here...", layout=widgets.Layout(height="150px", width="90%"))
btn_summarize = widgets.Button(description="Summarize Notes", button_style="success")

# Feature 3: Quiz Generator
quiz_text = widgets.Text(description="Topic:", placeholder="e.g., Photosynthesis")
btn_quiz = widgets.Button(description="Generate Quiz", button_style="info")

# Output Terminal Area
output_area = widgets.Output(layout=widgets.Layout(border="1px solid #ddd", padding="10px", margin="10px 0", min_height="150px"))

# Organize into Native Colab Tabs
tab_interface = widgets.Tab()
tab_interface.children = [
    widgets.VBox([widgets.HTML("<h3>💡 Simplify Difficult Topics</h3>"), concept_text, style_dropdown, btn_explain]),
    widgets.VBox([widgets.HTML("<h3>📝 Note Summarizer</h3>"), notes_text, btn_summarize]),
    widgets.VBox([widgets.HTML("<h3>🧠 On-Demand Quiz Generator</h3>"), quiz_text, btn_quiz])
]
tab_interface.set_title(0, 'Explain Concept')
tab_interface.set_title(1, 'Summarize Notes')
tab_interface.set_title(2, 'Quiz Generator')
```

```python
def on_explain_clicked(b):
    with output_area:
        clear_output()
        print("⏳ Analyzing and simplifying... Please wait...")
        prompt = f"Explain the concept of '{concept_text.value}' using a '{style_dropdown.value}' style. Make it clear, engaging, and easy to understand."
        response = get_ai_response(prompt, api_key_input.value)
        clear_output()
        print(response)

def on_summarize_clicked(b):
    with output_area:
        clear_output()
        print("⏳ Extracting core takeaways... Please wait...")
        prompt = f"Provide a clean structured summary of the following study notes. Break it down into 'Key Definitions' and 'Core Summary Points':\n\n{notes_text.value}"
        response = get_ai_response(prompt, api_key_input.value)
        clear_output()
        print(response)

def on_quiz_clicked(b):
    with output_area:
        clear_output()
        print("⏳ Crafting your custom quiz... Please wait...")
        prompt = f"Generate a 5-question multiple-choice quiz about '{quiz_text.value}'. For each question, provide 4 options (A, B, C, D). Include an 'Answer Key' at the very bottom with brief explanations."
        response = get_ai_response(prompt, api_key_input.value)
        clear_output()
        print(response)

# Link functions to buttons
btn_explain.on_click(on_explain_clicked)
btn_summarize.on_click(on_summarize_clicked)
btn_quiz.on_click(on_quiz_clicked)
```

```python
display(title_html)
display(widgets.VBox([api_key_input, help_text]))
display(tab_interface)
display(widgets.HTML("<br><h3>📢 Output Results:</h3>"))
display(output_area)
```
